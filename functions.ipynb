```python
#1. Difference between a function and a method in Python
Function: A block of reusable code. It is not bound to an object.

Method: A function associated with an object (usually a class).

Example:
def greet():          
    print("Hello")

name = "Alice"
print(name.upper())  
```

```python
#2. Function arguments and parameters in Python
Parameters: Variables listed in the function definition.

Arguments: Values passed to the function.

Example:

def add(a, b):  
    return a + b

add(2, 3)    
```

```python
#3. Ways to define and call a function in Python
Define using def or lambda.

Call by function_name(arguments).

Example:
def square(x):
    return x * x

print(square(5))

double = lambda x: x * 2
print(double(4)
```

```python
#4. Purpose of the return statement
It sends the result back to the caller and exits the function.

Example:

def multiply(a, b):
    return a * b

result = multiply(2, 3)
print(result)
```

```python
#5. Iterators vs Iterables
Iterable: An object which can be looped over (e.g., list, tuple).

Iterator: An object with __next__() method that gets next item.

Example:
nums = [1, 2, 3]      
it = iter(nums)       
print(next(it)) 
```

```python
#6. Generators in Python
Generators use yield to return values one at a time.

def countdown(n):
    while n > 0:
        yield n
        n -= 1

for val in countdown(3):
    print(val)
```

```python
#7. Advantages of generators
Memory efficient (lazy evaluation).

Faster for large datasets.

Useful for streaming data.

Example: Reading a large file line-by-line
```

```python
#8. Lambda function
An anonymous, one-line function.

Used for short operations, often in map, filter, etc.

Example:

square = lambda x: x**2
print(square(4))  
```

```python
#9. Purpose and usage of map()
Applies a function to every item of an iterable.

Example:
nums = [1, 2, 3]
squares = list(map(lambda x: x**2, nums))
print(squares) 
```

```python
#10. Difference between map(), reduce(), and filter()
Function	     Purpose	                                  Example
map()	   Applies function to all items	                map(f, iterable)
reduce()	Reduces iterable to a single value	        reduce(f, iterable)
filter()	Filters items based on condition	        filter(f, iterable)

Example:

from functools import reduce

nums = [1, 2, 3, 4]
print(list(map(lambda x: x*2, nums)))      # [2, 4, 6, 8]
print(list(filter(lambda x: x%2 == 0, nums)))  # [2, 4]
print(reduce(lambda x, y: x + y, nums))    # 10
```

```python
#1. Sum of all even numbers in a list

def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)
```

```python
# 2. Reverse a string

def reverse_string(s):
    return s[::-1]
```

```python
#3. Squares of a list of integers

def square_list(numbers):
    return [x**2 for x in numbers]
```

```python
#4. Check if a number is prime (1 to 200)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

for i in range(1, 201):
    print(f"{i}: {'Prime' if is_prime(i) else 'Not Prime'}")
```

```python
#5. Iterator class for Fibonacci sequence

class Fibonacci:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value
```

```python
#6. Generator for powers of 2

def powers_of_two(max_exp):
    for i in range(max_exp + 1):
        yield 2 ** i
```

```python
#7. Generator that reads a file line by line

def read_lines(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()
```

```python
#8. Sort list of tuples by second element using lambda

data = [(1, 3), (2, 1), (4, 5)]
sorted_data = sorted(data, key=lambda x: x[1])
```

```python
#9. Convert Celsius to Fahrenheit using map()
celsius = [0, 20, 30, 40]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
```

```python
#10. Remove vowels using filter()
def remove_vowels(s):
    return ''.join(filter(lambda x: x.lower() not in 'aeiou', s))
```
